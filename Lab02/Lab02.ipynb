## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:

- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# T7: Climate of five cities
tmax_shape = tmax.shape
annual_mean = tmax.mean(axis=1)
hottest_month_index = tmax.argmax(axis=1)
hottest_month = months[hottest_month_index]
five_city_monthly_mean = tmax.mean(axis=0)
months_above_35 = (tmax > 35).sum(axis=1)
lahore_fahrenheit = tmax[0] * 9/5 + 32
monsoon_rain = rain[:, 6:9]
monsoon_totals = monsoon_rain.sum(axis=1)
yearly_rain = rain.sum(axis=1)
monsoon_share = monsoon_totals / yearly_rain * 100
wettest_monsoon_index = monsoon_totals.argmax()

print("T7a: tmax shape:", tmax_shape)
print("T7b: Annual mean maximum temperature (°C):")
print(np.column_stack((cities, np.round(annual_mean, 2))))
print("T7c: Hottest month of each city:")
print(np.column_stack((cities, hottest_month, np.round(tmax[np.arange(len(cities)), hottest_month_index], 1))))
print("T7d: 5-city average temperature for each month (°C):")
print(np.column_stack((months, np.round(five_city_monthly_mean, 2))))
print("T7e: Number of months above 35 °C:")
print(np.column_stack((cities, months_above_35)))
print("T7f: Lahore temperatures in °F:")
print(np.column_stack((months, np.round(lahore_fahrenheit, 2))))
print("T7g: Monsoon rainfall, Jul-Sep (mm):")
print(np.column_stack((cities, np.round(monsoon_rain, 1))))
print("T7h: Monsoon rainfall total and yearly monsoon share:")
print(np.column_stack((cities, np.round(monsoon_totals, 1), np.round(monsoon_share, 2))))
print("T7h: Wettest monsoon city:", cities[wettest_monsoon_index])
print("T7 Interpretation: Islamabad's monsoon rainfall is a useful surprise because it is much wetter in Jul-Sep than Lahore in this dataset. Quetta is also unusual because its monsoon rainfall is very small compared with the other cities.")
```

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
`cpi` has 2 rows (FY2024, FY2025) and 10 columns (Jul to Apr), year-on-year CPI inflation in %.
- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.
- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# T8: Inflation FY2024 vs FY2025
average_inflation = cpi.mean(axis=1)
fy2024_above_25 = cpi_months[cpi[0] > 25]
inflation_drop = cpi[0] - cpi[1]
biggest_drop_index = inflation_drop.argmax()
fy2025_labels = np.where(cpi[1] < 5, "Low",
                         np.where(cpi[1] < 10, "Moderate", "High"))

print("T8a: Average inflation (%):")
print(np.column_stack((["FY2024", "FY2025"], np.round(average_inflation, 2))))
print("T8b: FY2024 months above 25%:", fy2024_above_25)
print("T8c: Inflation drop FY2024 - FY2025 for each month (percentage points):")
print(np.column_stack((cpi_months, np.round(inflation_drop, 2))))
print("T8d: Month with the biggest drop:", cpi_months[biggest_drop_index])
print("T8e: FY2025 labels (Low <5, Moderate 5-<10, High >=10):")
print(np.column_stack((cpi_months, fy2025_labels)))
print("T8 Interpretation: Inflation was much higher in FY2024 than in FY2025. The values fell strongly across the months, so prices were still rising in FY2025 but at a much slower rate.")
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9: Grace marks
grace_scores = np.clip(marks + 5, 0, 100)
pass_percentage = np.mean(grace_scores >= 50) * 100
passed_only_by_grace = np.where((marks < 50) & (grace_scores >= 50))[0]
z_scores = (grace_scores - grace_scores.mean()) / grace_scores.std()

print("T9a: Scores after +5 grace marks, capped at 100:", grace_scores)
print("T9b: Pass percentage after grace:", pass_percentage, "%")
print("T9c: Student indices who passed only because of grace:", passed_only_by_grace)
print("T9d: Z-scores after grace:")
print(np.round(z_scores, 3))
```
