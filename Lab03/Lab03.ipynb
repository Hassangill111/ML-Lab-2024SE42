### Task 2(h): Why the comparison fails

The `Price` column is stored as text (`object`) instead of a numeric data type.
The expression `df["Price"] > 1_000_000` compares strings with an integer.
Python cannot compare these different types in this way, so it raises a `TypeError`.

The price values must first be cleaned by removing text such as `Rs` and commas,
then converted to a numeric type. This is performed in Task 3.

## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# Task 3: Clean in the required order
# (a) Missing values
print("T3a missing values before cleaning:")
display(df.isna().sum())
rows_before_missing = len(df)
clean = df.dropna(subset=["Brand", "Model"]).copy()
rows_removed_missing = rows_before_missing - len(clean)
for col in ["Registered City", "Fuel", "Transaction Type"]:
    clean[col] = clean[col].fillna("Unknown")
print("T3a: rows removed for missing Brand/Model =", rows_removed_missing)

# (b) Formatting: normalize city text and remove price formatting
clean["Registered City"] = clean["Registered City"].astype("string").str.strip().str.title()
clean["Price"] = (clean["Price"].astype(str)
                  .str.replace("Rs", "", regex=False)
                  .str.replace(",", "", regex=False)
                  .str.strip())

# (c) Data types
clean["Price"] = pd.to_numeric(clean["Price"], errors="raise").astype("int64")
clean["Year"] = clean["Year"].astype("Int64")
print("T3c: Price dtype =", clean["Price"].dtype, "| Year dtype =", clean["Year"].dtype)

# (d) Inconsistent categories
cities_before = clean["Registered City"].nunique(dropna=True)
clean["Registered City"] = clean["Registered City"].replace({"Sukkar": "Sukkur"})
clean["Model"] = clean["Model"].replace({"Corrolla Altis": "Corolla Altis"})
cities_after = clean["Registered City"].nunique(dropna=True)
print("T3d: different cities before =", cities_before, "| after =", cities_after)

# (e) Duplicates (after text normalization)
duplicate_count = int(clean.duplicated().sum())
clean = clean.drop_duplicates().copy()
print("T3e: exact duplicate ads removed =", duplicate_count)
print("T3e: rows after duplicate removal =", len(clean))

# (f) Invalid values
invalid_km = ((clean["KMs Driven"] < 100) |
              (clean["KMs Driven"] > 1_000_000) |
              (clean["KMs Driven"].isin([12345, 123456])))
clean.loc[invalid_km, "KMs Driven"] = np.nan
fake_price = clean["Price"].isin([12_345_678, 87_654_321])
implausible_brand_price = clean["Brand"].isin(["Suzuki", "Daihatsu"]) & (clean["Price"] > 5_000_000)
invalid_price_rows = fake_price | implausible_brand_price
print("T3f: invalid-price ads removed =", int(invalid_price_rows.sum()))
clean = clean.loc[~invalid_price_rows].copy()

# (g) Required self-check
print("T3g: final cleaned shape =", clean.shape)
assert clean.shape == (20484, 9), f"Expected (20484, 9), got {clean.shape}"
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# Task 4: Analyse

analysis = clean.copy()

# T4a: Add Price_lakh column
analysis["Price_lakh"] = analysis["Price"] / 100_000

print("T4a: Price_lakh (Price divided by 100,000)")
display(analysis[["Price", "Price_lakh"]].head())


# T4b: Add Age column
analysis["Age"] = 2020 - analysis["Year"]

print("T4b: Age (2020 - Year)")
display(analysis[["Year", "Age"]].head())


# T4c: Five most common brands
print("T4c: Five most common brands")
display(analysis["Brand"].value_counts().head(5))


# T4d: Five cities with the most advertisements
print("T4d: Five cities with the most advertisements")
display(analysis["Registered City"].value_counts().head(5))


# T4e: Median price in lakh by fuel type, highest first
print("T4e: Median Price_lakh by fuel type")
fuel_median = (
    analysis.groupby("Fuel")["Price_lakh"]
    .median()
    .sort_values(ascending=False)
)

display(fuel_median)


# T4f: Median Corolla GLI price for each year from 2010 to 2018
print("T4f: Corolla GLI median price by year (2010–2018)")

gli = analysis.loc[
    analysis["Model"].astype("string").str.contains(
        "Corolla GLI", case=False, na=False
    )
    & analysis["Year"].between(2010, 2018)
]

gli_year = (
    gli.groupby("Year")["Price_lakh"]
    .median()
    .reindex(range(2010, 2019))
)

display(gli_year)

# Compare the first and last available years
available = gli_year.dropna()

if len(available) >= 2:
    first_year = available.index[0]
    last_year = available.index[-1]
    first_price = available.iloc[0]
    last_price = available.iloc[-1]

    change_per_year = (
        (last_price - first_price) / (last_year - first_year)
    )

    print(
        f"T4f: Average median price change = "
        f"{change_per_year:.3f} lakh rupees per year"
    )
    print(
        f"T4f: From {first_year} ({first_price:.2f} lakh) "
        f"to {last_year} ({last_price:.2f} lakh)"
    )
else:
    print("T4f: Not enough available year groups to calculate the change.")


# T4g: Five most expensive advertisements
print("T4g: Five most expensive advertisements")

expensive_ads = analysis.nlargest(
    5, "Price"
)[["Brand", "Model", "Year", "Price", "Registered City"]]

display(expensive_ads)


# T4h: Cash versus Installment/Leasing
print("T4h: Advertisement count and median price")

transaction_summary = (
    analysis[
        analysis["Transaction Type"].isin(
            ["Cash", "Installment/Leasing"]
        )
    ]
    .groupby("Transaction Type")["Price"]
    .agg(["count", "median"])
)

display(transaction_summary)
```
